# Лабораторная работа: Генерация синтетических данных (Вариант 8)
Импорт необходимых библиотек и фиксация параметров.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs

# Начальные параметры
np.random.seed(0)
n_samples = 100
random_state = 3

# Параметры Варианта 8
centers = [(-0.3, 0.3), (0.7, 0.7)]
cluster_std = [(0.07, 0.5), (0.06, 0.5)]
```

## Генерация данных и построение графика

```python
# Генерация данных c помощью make_blobs
X, y = make_blobs(
    n_samples=n_samples,
    centers=centers,
    cluster_std=cluster_std,
    random_state=random_state
)

# Построение точечного графика
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='tab10', edgecolors='k')
plt.title('График кластеров (Вариант 8)')
plt.xlabel('X1')
plt.ylabel('X2')
plt.grid(True)
plt.savefig('img/blobs_variant8.png', dpi=300, bbox_inches='tight')
plt.show()
```
